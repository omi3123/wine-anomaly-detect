# Wine Anomaly Detection

Isolation Forest vs Local Outlier Factor with PCA visualization.

```

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
```

## Load Data

```

data = load_wine()
X = data.data
df = pd.DataFrame(X, columns=data.feature_names)
df.head()
```

## Preprocessing

```

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## PCA for Visualization

```

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.figure(dpi=150)
plt.scatter(X_pca[:,0], X_pca[:,1])
plt.title("PCA Projection")
plt.show()
```

## Isolation Forest

```

iso = IsolationForest(contamination=0.05, random_state=42)
iso_labels = iso.fit_predict(X_scaled)

plt.figure(dpi=150)
plt.scatter(X_pca[:,0], X_pca[:,1], c=iso_labels)
plt.title("Isolation Forest Anomalies")
plt.show()
```

## Local Outlier Factor

```

lof = LocalOutlierFactor(n_neighbors=20, contamination=0.05)
lof_labels = lof.fit_predict(X_scaled)

plt.figure(dpi=150)
plt.scatter(X_pca[:,0], X_pca[:,1], c=lof_labels)
plt.title("LOF Anomalies")
plt.show()
```

## Conclusion

```

print("Isolation Forest and LOF successfully detected anomalies. PCA helped visualize them clearly.")
```
